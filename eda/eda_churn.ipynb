**Objetivos**
- Remover valores faltantes<br>

```python
df['TotalCharges'] = df['TotalCharges'].replace(' ', np.nan)
df['TotalCharges'] = pd.to_numeric(df['TotalCharges'], errors='coerce')
df['TotalCharges'].head()
```

```python
cols = df.select_dtypes(include='number').columns

cols = [
    col for col in cols
    if not df[col].dropna().isin([0, 1]).all()
]

n_colunas = 3
n_linhas = math.ceil(len(cols) / n_colunas)

fig = make_subplots(
    rows=n_linhas,
    cols=n_colunas,
    subplot_titles=cols
)

for i, col in enumerate(cols):
    linha = i // n_colunas + 1
    coluna = i % n_colunas + 1

    fig.add_trace(
        go.Box(
            y=df[col],
            name=col,
            showlegend=False
        ),
        row=linha,
        col=coluna
    )

fig.update_layout(
    height=400 * n_linhas,
    width=1000,
    showlegend=False
)

fig.show()
```

```python
gen_churn_to = df.groupby('gender')['Churn'].count().reset_index(name='count')

fig = px.bar(gen_churn_to, x="gender", y="count")

fig.update_layout(
    title="Total por gênero",
    xaxis_title="Gênero",
    yaxis_title="Quantidade",
    width=500,
    height=500
)

fig.update_traces(
    texttemplate="%{y}",
    textposition="outside"
)

fig.show()
```

```python
# Porcentagem de cancelamento por gênero
homens = gen_churn_to[gen_churn_to['gender'] == 'Male']['count'].values[0]
mulheres = gen_churn_to[gen_churn_to['gender'] == 'Female']['count'].values[0]

homens_cancelados = gen_churn[gen_churn['gender'] == 'Male']['count'].values[1]
mulheres_canceladas = gen_churn[gen_churn['gender'] == 'Female']['count'].values[1]

homens_per = homens_cancelados / homens * 100
mulheres_per = mulheres_canceladas / mulheres * 100

print("Porcentagem de cancelamento por gênero:")
print(f"Homens: {homens_per:.2f}%")
print(f"Mulheres: {mulheres_per:.2f}%")
print(f"Diferenca: {(mulheres_per - homens_per):.2f}")
```

**Resultado**<br>
Cancelamento por Gênero:
- De 3555 **Homens**, 930 cancelaram -> 26.16%
- De 3488 **Mulheres**, 939 cancelaram -> 26.92%
- Diferença -> **0.76%**

**Insight**
> **Gênero não apresentou uma diferença significativa para a chance de cancelamento.**

## **2. Insight**

**Váriavel**
> Quantiadade de serviços de clientes que cancelaram

**Pergunta**
> Quantidades de servicos com maiores números de clientes que cancelaram?

**Análise**

```python
clientes_cancelaram = df[df['Churn'] == 'Yes'].copy()
colunas_texto = clientes_cancelaram.select_dtypes(include=['object', 'string']).columns

clientes_cancelaram[colunas_texto] = df[colunas_texto].replace({'Yes': 1, 'No': 0})
clientes_cancelaram.head(5)
```

```python
colunas_servicos_reais = [
    'PhoneService', 'MultipleLines', 'OnlineSecurity',
    'OnlineBackup', 'DeviceProtection', 'TechSupport',
    'StreamingTV', 'StreamingMovies'
]

servicos_numericos = clientes_cancelaram[colunas_servicos_reais].apply(pd.to_numeric, errors='coerce').fillna(0)
clientes_cancelaram['total_servicos'] = servicos_numericos.sum(axis=1)

# Agrupando os dados para contar a quantidade de clientes por número de serviços contratados
cli = clientes_cancelaram[clientes_cancelaram['total_servicos'] > 0].copy()
servicos_agrupados = cli['total_servicos'].value_counts().reset_index(name='count')
servicos_agrupados.columns = ['total_servicos', 'count']
servicos_agrupados = servicos_agrupados.sort_values(by='total_servicos')

# Criando o gráfico de barras
fig = px.bar(
    servicos_agrupados,
    x="total_servicos",
    y="count",
    text_auto=True,
    opacity=0.75
)

fig.update_layout(
    title="Total de serviços de clientes que cancelaram",
    xaxis_title="Quantidade de Serviços Contratados",
    yaxis_title="Quantidade de Clientes (Contagem)",
    width=1000,
    height=500
)

fig.update_traces(
    textposition="outside"
)

fig.show()
```

**Resultado**<br>
De 1 - 3 serviços têm a maior quantidade dos clientes que cancelaram.

**Insight**
> **Clientes que cancelaram têm a tendência de terem menos serviços solicitados.**

## **3. Insight**

**Váriavel**
> Valor mensal com valor total pago

**Pergunta**
> Tem relação entre valor mensal e valor total pago de clientes que cancelaram e dos que permaneceram?

**Análise**

```python
df_analis = df[['MonthlyCharges', 'TotalCharges','Churn']].copy()
colunas = ['MonthlyCharges', 'TotalCharges']

scaler = MinMaxScaler()

df_analis[colunas] = scaler.fit_transform(df_analis[colunas])
df_analis.head()
```

```python
fig = px.scatter(
    df_analis,
    x="MonthlyCharges",
    y="TotalCharges",
    color="Churn",
    color_discrete_map={
        "Yes": "red",
        "No": "green"
    },
    opacity=0.4,
    labels={
        "MonthlyCharges": "Valor Mensal",
        "TotalCharges": "Total Pago",
        "Churn": "Cancelou"
    },
    title="Relação entre Valor Mensal e Total Pago"
)

fig.update_layout(
    width=800,
    height=800
)


fig.show()
```

```python
df_yes = df_analis[df_analis['Churn'] == 'Yes'].copy()

print("Correlação de valor mensal com total pago de clientes que fizeram cancelamento\n")
print(df_yes[colunas].corr())
```

```python
df_yes = df_analis[df_analis['Churn'] == 'No'].copy()

print("Correlação de valor mensal com total pago que não fizeram cancelamento\n")
print(df_yes[colunas].corr())
```

**Resultado**<br>
Pelo gráfico podemos observar que os valores parecem estar bem dispersos, mas quando fazemos o cálculo da correlação:
- Para clientes que **não fizeram cancelamento** -> **Correlação de 0.76** aproximadamente, com uma **correlação positiva forte**.
- Para cliente que **fizeram cancelamento** -> **Correlação de 0.54** aproximadamente, com uma **correlação positiva moderada**.

**Insight**
> **Esse comportamento sugere que a relação entre o valor mensal e o total pago varia de acordo com o grupo de cancelamento, mas que essa relação é mais forte em relação aos que não cancelaram.**
